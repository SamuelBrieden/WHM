```python
def slopefinder(kv, Pv1, Pv2, iguess):
    slopediff = np.gradient(np.log10(Pv1),np.log10(kv)) - np.gradient(np.log10(Pv2),np.log10(kv))
    guess = slopediff[iguess]
    if guess==0.0:
        return kguess
    elif guess>0.0:
        i=0
        while guess>0.0:
            i+=1
            guess = np.interp(kv[iguess-i],kv,slopediff)    
        return kv[iguess-i]
    else:
        i=0
        while guess<0.0:
            i+=1
            guess = np.interp(kv[iguess+i],kv,slopediff)    
        return kv[iguess-i]

def slopefinder_new(kv, Pv1, Pv2):
    slopediff = np.gradient(np.log10(Pv1),np.log10(kv)) - np.gradient(np.log10(Pv2),np.log10(kv))
    signs = np.sign(slopediff[0:len(kv)-3])
    signchange = ((np.roll(signs, 1) - signs) != 0).astype(int)
    indices = np.where(signchange==1)  
    return kv[indices[0][-1]-1]

def concatenate_vectors(x, yleft, yright, xcut):
    N = len(yleft)
    result = np.zeros(N)
    leftindices = np.where(x<=xcut)
    rightindices = np.where(x>xcut)
    result[leftindices[0]] = yleft[leftindices[0]]
    result[rightindices[0]] = yleft[rightindices[0][0]]/yright[rightindices[0][0]]*yright[rightindices[0]]
    return result
    
```
